```python
import torch
print(torch.__version__)
```

```python
if torch.cuda.is_available():
  print("GPu is available")
else:
  print("Using CPU")
```

```python
#Ways to Create a tensor
```

```python
#Using Empty
torch.empty(2,3) #creates a space in the memory and whatever exising value was sitting there prints it out for us
```

```python
#Using Zeros
torch.zeros(2,3)
```

```python
#using rand
torch.rand(2,3) #creates a tensor where all the values are between 0 and 1
```

```python
#using tensor
torch.tensor([[1,2,3], [4,5,6]]) #creates a tensor with the values we want it to have
```

```python
#using arange
torch.arange(0,10,2)
```

```python
#using eye (identity matrix)
torch.eye(5) #creates a 5x5 identity matrix
```

```python
#Tensor Shapes
x = torch.tensor([[1,2,3], [4,5,6]])
x
```

```python
x.shape
```

```python
#if we want to create a similar tensor out of another tensor we can use methods like
torch.empty_like(x)
```

```python
torch.zeros_like(x)
```

```python
#Tensor Datatypes
x.dtype
```

```python
torch.tensor([1,2,3], dtype=torch.float64)
```

```python
torch.tensor([1.0,2.0,3.0], dtype=torch.int64)
```

```python
x.to(torch.float64)
```
